```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from factor_analyzer import FactorAnalyzer
from sklearn.preprocessing import StandardScaler

# Assuming df_factor is your dataframe
# If it's not defined, you'll need to load or create it first

# Step 1: Prepare the data
scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df_factor2), columns=df_factor2.columns)

# Step 2: Perform Factor Analysis with Oblimin rotation
n_factors = len(df_factor2.columns)  # Start with the number of columns as the number of factors
fa_oblimin = FactorAnalyzer(rotation='oblimin', n_factors=n_factors)
fa_oblimin.fit(df_scaled)

# Step 3: Get and print eigenvalues
eigenvalues, _ = fa_oblimin.get_eigenvalues()
print("Eigenvalues:")
for i, ev in enumerate(eigenvalues):
    print(f"Factor {i+1}: {ev:.4f}")

# Determine significant factors (eigenvalues > 1)
significant_factors = sum(eigenvalues > 0.5)
print(f"\nNumber of significant factors (eigenvalue > 1): {significant_factors}")

# Step 4: Get factor loadings (for all factors)
oblimin_loadings = pd.DataFrame(fa_oblimin.loadings_, 
                                columns=[f'Factor{i+1}' for i in range(n_factors)], 
                                index=df_factor2.columns)

# Step 5: Print Oblimin factor loadings for significant factors
print("\nOblimin Factor Loadings (for significant factors):")
print(oblimin_loadings.iloc[:, :significant_factors])

# Step 6: Calculate and print Oblimin factors
oblimin_factors = fa_oblimin.transform(df_scaled)
oblimin_factors_df = pd.DataFrame(oblimin_factors[:, :significant_factors], 
                                  columns=[f'Factor{i+1}' for i in range(significant_factors)])
print("\nOblimin Factors:")
print(oblimin_factors_df.head())  # Print first 5 rows

# Step 7: Print factor correlations for significant factors
factor_correlations = fa_oblimin.corr_
print("\nFactor Correlations:")
print(pd.DataFrame(factor_correlations[:significant_factors, :significant_factors], 
                   columns=[f'Factor{i+1}' for i in range(significant_factors)],
                   index=[f'Factor{i+1}' for i in range(significant_factors)]))

# Step 8: Plot Oblimin Factor Loadings for significant factors
plt.figure(figsize=(15, 8))
ax = plt.gca()
oblimin_loadings.iloc[:, :significant_factors].plot(kind='bar', ax=ax, width=0.8)
plt.title('Oblimin Factor Loadings')
plt.xlabel('Variables')
plt.ylabel('Loading')
plt.legend(title='Factors', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
# Save the figure
#plt.savefig('analysis_final/graphs/oblimin_factor_loading.svg', format='svg')
plt.show()

# Step 9: Create Scree plot
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(eigenvalues) + 1), eigenvalues, 'bo-')
plt.title('Scree Plot')
plt.xlabel('Factor')
plt.ylabel('Eigenvalue')
#plt.savefig('analysis_final/graphs/oblimin_scree.svg', format='svg')
#plt.axhline(y=1, color='r', linestyle='--')  # Add a horizontal line at y=1
plt.show()
```
