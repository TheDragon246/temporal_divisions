## Correlation between DD/DS contrast and questionnaires

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr

# Define the columns of interest
columns_of_interest = [
    'object_view_rt_DD-DS',
    'object_view_acc_DS-DD',
    'mapping_dist_DD-DS',
    'dist_est_corr_DS-DD',
    'dist_est_DD-DS',
    'dist_comp_DS-DD',
    'jrd_lr_DS-DD',
    'jrd_diff_DD-DS',
    'sbsod',
    'nsq',
    'vgq'
]

# Create column renaming dictionary
column_rename_dict = {
    'object_view_rt_DD-DS': 'Object viewing RT',
    'object_view_acc_DS-DD': 'Object viewing Accuracy',
    'mapping_dist_DD-DS': 'Mapping',
    'dist_est_corr_DS-DD': 'Distance estimation Correlation',
    'dist_est_DD-DS': 'Distance estimation Elongation',
    'dist_comp_DS-DD': 'Distance comparison',
    'jrd_lr_DS-DD': 'JRD left-right accuracy',
    'jrd_diff_DD-DS': 'JRD angular accuracy',
    'sbsod': 'SBSOD',
    'nsq': 'NSQ',
    'vgq': 'VGQ'
}

# Create a subset dataframe and rename columns
df_subset = df_factor[columns_of_interest].copy()
df_renamed = df_subset.rename(columns=column_rename_dict)

# Define the desired order for the matrix
desired_order = [
    'JRD angular accuracy',
    'JRD left-right accuracy', 
    'Distance estimation Elongation',
    'Distance estimation Correlation',
    'Distance comparison',
    'Object viewing RT',
    'Object viewing Accuracy',
    'Mapping',
    'SBSOD',
    'NSQ',
    'VGQ'
]

# Create correlation matrix with renamed columns
correlation_matrix = df_renamed.corr()

# Reorder the correlation matrix according to desired order
correlation_matrix = correlation_matrix.reindex(index=desired_order, columns=desired_order)

# Function to calculate p-values for correlations
def calculate_pvalues(df):
    """
    Calculate p-values for correlations between all pairs of variables
    """
    cols = df.columns
    pvalues = np.zeros((len(cols), len(cols)))
    
    for i, col1 in enumerate(cols):
        for j, col2 in enumerate(cols):
            if i == j:
                pvalues[i, j] = 0  # Perfect correlation with itself
            else:
                # Remove NaN values for pair-wise correlation
                valid_data = df[[col1, col2]].dropna()
                if len(valid_data) > 2:  # Need at least 3 points for correlation
                    _, p_val = pearsonr(valid_data[col1], valid_data[col2])
                    pvalues[i, j] = p_val
                else:
                    pvalues[i, j] = np.nan
    
    return pd.DataFrame(pvalues, columns=cols, index=cols)

# Calculate p-values with renamed columns and reorder
pvalue_matrix = calculate_pvalues(df_renamed)
pvalue_matrix = pvalue_matrix.reindex(index=desired_order, columns=desired_order)

# Display the correlation matrix
print("Correlation Matrix (DD-DS Analysis):")
print(correlation_matrix)
print("\nP-value Matrix:")
print(pvalue_matrix)

# Create a heatmap visualization with renamed columns
plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, 
            annot=True,  # Show correlation values
            cmap='coolwarm',  # Color scheme
            center=0,  # Center colormap at 0
            square=True,  # Make cells square
            fmt='.3f',  # Format to 3 decimal places
            cbar_kws={'label': 'Correlation Coefficient'},
            annot_kws={'size': 9})  # Smaller annotation size for readability
plt.title('ND-NS Correlation Matrix', fontsize=16, fontweight='bold')
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.tight_layout()

# Save as SVG file
plt.savefig('correlation_matrix_NDNS_renamed_heatmap.svg', format='svg', dpi=300, bbox_inches='tight')
plt.show()

# Create significance heatmap with renamed columns
plt.figure(figsize=(14, 12))
# Create a mask for significance (p < 0.05)
significance_mask = pvalue_matrix >= 0.05
sns.heatmap(correlation_matrix, 
            annot=True,
            mask=significance_mask,  # Mask non-significant correlations
            cmap='coolwarm',
            center=0,
            square=True,
            fmt='.3f',
            cbar_kws={'label': 'Correlation Coefficient'},
            annot_kws={'size': 9})  # Smaller annotation size for readability
plt.title('ND-NS Statistically Significant Correlations (p < 0.05)', fontsize=16, fontweight='bold')
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.tight_layout()
plt.savefig('significant_correlations_NDNS_renamed_heatmap.svg', format='svg', dpi=300, bbox_inches='tight')
plt.show()

# Optional: Get just the upper triangle (since correlation matrix is symmetric)
upper_triangle = correlation_matrix.where(
    np.triu(np.ones_like(correlation_matrix, dtype=bool), k=1)
)
print("\nUpper Triangle of Correlation Matrix:")
print(upper_triangle)

# Find strong correlations (absolute value > 0.7)
print("\n" + "="*60)
print("STRONG CORRELATIONS ANALYSIS (DD-DS)")
print("="*60)

strong_correlations = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]
        p_value = pvalue_matrix.iloc[i, j]
        if abs(corr_value) > 0.7:
            strong_correlations.append({
                'Variable 1': correlation_matrix.columns[i],
                'Variable 2': correlation_matrix.columns[j],
                'Correlation': corr_value,
                'P-value': p_value,
                'Significant': p_value < 0.05
            })

if strong_correlations:
    print(f"\nStrong correlations (|r| > 0.7):")
    for corr in strong_correlations:
        significance = "***" if corr['P-value'] < 0.001 else "**" if corr['P-value'] < 0.01 else "*" if corr['P-value'] < 0.05 else "ns"
        print(f"{corr['Variable 1']} <-> {corr['Variable 2']}: r = {corr['Correlation']:.3f}, p = {corr['P-value']:.4f} {significance}")
else:
    print("\nNo strong correlations (|r| > 0.7) found.")

# Find statistically significant correlations
print("\n" + "="*60)
print("STATISTICALLY SIGNIFICANT CORRELATIONS ANALYSIS (DD-DS)")
print("="*60)

significant_correlations = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]
        p_value = pvalue_matrix.iloc[i, j]
        if p_value < 0.05 and not np.isnan(p_value):
            significant_correlations.append({
                'Variable 1': correlation_matrix.columns[i],
                'Variable 2': correlation_matrix.columns[j],
                'Correlation': corr_value,
                'P-value': p_value,
                'Strength': 'Strong' if abs(corr_value) > 0.7 else 'Moderate' if abs(corr_value) > 0.3 else 'Weak'
            })

# Sort by absolute correlation value (strongest first)
significant_correlations.sort(key=lambda x: abs(x['Correlation']), reverse=True)

if significant_correlations:
    print(f"\nStatistically significant correlations (p < 0.05):")
    print("-" * 80)
    for corr in significant_correlations:
        significance_level = "***" if corr['P-value'] < 0.001 else "**" if corr['P-value'] < 0.01 else "*"
        print(f"{corr['Variable 1']} <-> {corr['Variable 2']}: "
              f"r = {corr['Correlation']:+.3f}, p = {corr['P-value']:.4f} {significance_level} "
              f"({corr['Strength']})")
else:
    print("\nNo statistically significant correlations (p < 0.05) found.")

# Summary statistics
print("\n" + "="*60)
print("DD-DS ANALYSIS SUMMARY")
print("="*60)
total_pairs = len(correlation_matrix.columns) * (len(correlation_matrix.columns) - 1) // 2
print(f"Total variable pairs analyzed: {total_pairs}")
print(f"Strong correlations (|r| > 0.7): {len(strong_correlations)}")
print(f"Statistically significant correlations (p < 0.05): {len(significant_correlations)}")

# Identify difference score correlations specifically (using original column names for logic)
print(f"\nDifference Score Analysis:")
diff_score_vars = [column_rename_dict[col] for col in columns_of_interest if 'DD-DS' in col or 'DS-DD' in col]
questionnaire_vars = ['SBSOD', 'NSQ', 'VGQ']
print(f"Difference score variables: {len(diff_score_vars)}")
print(f"Questionnaire variables: {len(questionnaire_vars)}")

# Correlations between difference scores and questionnaires
diff_quest_correlations = []
for diff_var in diff_score_vars:
    for quest_var in questionnaire_vars:
        if diff_var in correlation_matrix.index and quest_var in correlation_matrix.columns:
            corr_val = correlation_matrix.loc[diff_var, quest_var]
            p_val = pvalue_matrix.loc[diff_var, quest_var]
            if p_val < 0.05 and not np.isnan(p_val):
                diff_quest_correlations.append({
                    'Difference Score': diff_var,
                    'Questionnaire': quest_var,
                    'Correlation': corr_val,
                    'P-value': p_val
                })

if diff_quest_correlations:
    print(f"\nSignificant correlations between difference scores and questionnaires:")
    for corr in sorted(diff_quest_correlations, key=lambda x: abs(x['Correlation']), reverse=True):
        significance_level = "***" if corr['P-value'] < 0.001 else "**" if corr['P-value'] < 0.01 else "*"
        print(f"{corr['Difference Score']} <-> {corr['Questionnaire']}: "
              f"r = {corr['Correlation']:+.3f}, p = {corr['P-value']:.4f} {significance_level}")

# Significance levels legend
print(f"\nSignificance levels:")
print(f"*** p < 0.001")
print(f"**  p < 0.01")
print(f"*   p < 0.05")
print(f"ns  not significant")

# Create a comprehensive results dataframe
if significant_correlations:
    results_df = pd.DataFrame(significant_correlations)
    print(f"\nDetailed Results Table:")
    print(results_df.to_string(index=False))
    
    # Optionally save results
    # results_df.to_csv('significant_correlations_DDDS_renamed_results.csv', index=False)
    # print(f"\nResults saved to 'significant_correlations_DDDS_renamed_results.csv'")

# Export correlation matrix and p-values to CSV if needed
# correlation_matrix.to_csv('correlation_matrix_DDDS_renamed.csv')
# pvalue_matrix.to_csv('pvalue_matrix_DDDS_renamed.csv')
# print(f"\nMatrices saved to CSV files")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr

# Define the columns of interest
columns_of_interest = [
    'object_view_rt',
    'object_view_acc',
    'mapping_dist',
    'dist_est_corr',
    'dist_est',
    'dist_comp',
    'jrd_lr',
    'jrd_diff',
    'sbsod',
    'nsq',
    'vgq'
]

# Create correlation matrix
correlation_matrix = df_factor[columns_of_interest].corr()

# Function to calculate p-values for correlations
def calculate_pvalues(df):
    """
    Calculate p-values for correlations between all pairs of variables
    """
    cols = df.columns
    pvalues = np.zeros((len(cols), len(cols)))
    
    for i, col1 in enumerate(cols):
        for j, col2 in enumerate(cols):
            if i == j:
                pvalues[i, j] = 0  # Perfect correlation with itself
            else:
                # Remove NaN values for pair-wise correlation
                valid_data = df[[col1, col2]].dropna()
                if len(valid_data) > 2:  # Need at least 3 points for correlation
                    _, p_val = pearsonr(valid_data[col1], valid_data[col2])
                    pvalues[i, j] = p_val
                else:
                    pvalues[i, j] = np.nan
    
    return pd.DataFrame(pvalues, columns=cols, index=cols)

# Calculate p-values
pvalue_matrix = calculate_pvalues(df_factor[columns_of_interest])

# Display the correlation matrix
print("Correlation Matrix:")
print(correlation_matrix)
print("\nP-value Matrix:")
print(pvalue_matrix)

# Create a heatmap visualization
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, 
            annot=True,  # Show correlation values
            cmap='coolwarm',  # Color scheme
            center=0,  # Center colormap at 0
            square=True,  # Make cells square
            fmt='.3f',  # Format to 3 decimal places
            cbar_kws={'label': 'Correlation Coefficient'})
plt.title('Correlation Matrix Heatmap', fontsize=16, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()

# Save as SVG file
#plt.savefig('correlation_matrix_overall_q_heatmap.svg', format='svg', dpi=300, bbox_inches='tight')
plt.show()

# Create significance heatmap
plt.figure(figsize=(12, 10))
# Create a mask for significance (p < 0.05)
significance_mask = pvalue_matrix >= 0.05
sns.heatmap(correlation_matrix, 
            annot=True,
            mask=significance_mask,  # Mask non-significant correlations
            cmap='coolwarm',
            center=0,
            square=True,
            fmt='.3f',
            cbar_kws={'label': 'Correlation Coefficient'})
plt.title('Statistically Significant Correlations (p < 0.05)', fontsize=16, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
#plt.savefig('significant_correlations_heatmap.svg', format='svg', dpi=300, bbox_inches='tight')
plt.show()

# Optional: Get just the upper triangle (since correlation matrix is symmetric)
upper_triangle = correlation_matrix.where(
    np.triu(np.ones_like(correlation_matrix, dtype=bool), k=1)
)
print("\nUpper Triangle of Correlation Matrix:")
print(upper_triangle)

# Find strong correlations (absolute value > 0.7)
print("\n" + "="*60)
print("STRONG CORRELATIONS ANALYSIS")
print("="*60)

strong_correlations = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]
        p_value = pvalue_matrix.iloc[i, j]
        if abs(corr_value) > 0.7:
            strong_correlations.append({
                'Variable 1': correlation_matrix.columns[i],
                'Variable 2': correlation_matrix.columns[j],
                'Correlation': corr_value,
                'P-value': p_value,
                'Significant': p_value < 0.05
            })

if strong_correlations:
    print(f"\nStrong correlations (|r| > 0.7):")
    for corr in strong_correlations:
        significance = "***" if corr['P-value'] < 0.001 else "**" if corr['P-value'] < 0.01 else "*" if corr['P-value'] < 0.05 else "ns"
        print(f"{corr['Variable 1']} <-> {corr['Variable 2']}: r = {corr['Correlation']:.3f}, p = {corr['P-value']:.4f} {significance}")
else:
    print("\nNo strong correlations (|r| > 0.7) found.")

# Find statistically significant correlations
print("\n" + "="*60)
print("STATISTICALLY SIGNIFICANT CORRELATIONS ANALYSIS")
print("="*60)

significant_correlations = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]
        p_value = pvalue_matrix.iloc[i, j]
        if p_value < 0.05 and not np.isnan(p_value):
            significant_correlations.append({
                'Variable 1': correlation_matrix.columns[i],
                'Variable 2': correlation_matrix.columns[j],
                'Correlation': corr_value,
                'P-value': p_value,
                'Strength': 'Strong' if abs(corr_value) > 0.7 else 'Moderate' if abs(corr_value) > 0.3 else 'Weak'
            })

# Sort by absolute correlation value (strongest first)
significant_correlations.sort(key=lambda x: abs(x['Correlation']), reverse=True)

if significant_correlations:
    print(f"\nStatistically significant correlations (p < 0.05):")
    print("-" * 80)
    for corr in significant_correlations:
        significance_level = "***" if corr['P-value'] < 0.001 else "**" if corr['P-value'] < 0.01 else "*"
        print(f"{corr['Variable 1']} <-> {corr['Variable 2']}: "
              f"r = {corr['Correlation']:+.3f}, p = {corr['P-value']:.4f} {significance_level} "
              f"({corr['Strength']})")
else:
    print("\nNo statistically significant correlations (p < 0.05) found.")

# Summary statistics
print("\n" + "="*60)
print("SUMMARY")
print("="*60)
total_pairs = len(correlation_matrix.columns) * (len(correlation_matrix.columns) - 1) // 2
print(f"Total variable pairs analyzed: {total_pairs}")
print(f"Strong correlations (|r| > 0.7): {len(strong_correlations)}")
print(f"Statistically significant correlations (p < 0.05): {len(significant_correlations)}")

# Significance levels legend
print(f"\nSignificance levels:")
print(f"*** p < 0.001")
print(f"**  p < 0.01")
print(f"*   p < 0.05")
print(f"ns  not significant")

# Create a comprehensive results dataframe
if significant_correlations:
    results_df = pd.DataFrame(significant_correlations)
    print(f"\nDetailed Results Table:")
    print(results_df.to_string(index=False))
    
    # Optionally save results
    # results_df.to_csv('significant_correlations_results.csv', index=False)
    # print(f"\nResults saved to 'significant_correlations_results.csv'")

# Export correlation matrix and p-values to CSV if needed
# correlation_matrix.to_csv('correlation_matrix.csv')
# pvalue_matrix.to_csv('pvalue_matrix.csv')
# print(f"\nMatrices saved to CSV files")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# Define the task pairs to correlate
task_pairs = [
    ('object_view_rt', 'object_view_rt_DD-DS'),
    ('object_view_acc', 'object_view_acc_DS-DD'),
    ('mapping_dist', 'mapping_dist_DD-DS'),
    ('dist_est_corr', 'dist_est_corr_DS-DD'),
    ('dist_est', 'dist_est_DD-DS'),
    ('dist_comp', 'dist_comp_DS-DD'),
    ('jrd_lr', 'jrd_lr_DS-DD'),
    ('jrd_diff', 'jrd_diff_DD-DS')
]

# Create a results DataFrame to store correlations
correlation_results = []

# Calculate correlations for each pair
for overall_task, diff_task in task_pairs:
    # Remove any rows with missing values for both columns
    valid_data = df_factor[[overall_task, diff_task]].dropna()
    
    if len(valid_data) > 2:  # Need at least 3 data points for correlation
        # Calculate Pearson correlation
        corr_coef, p_value = pearsonr(valid_data[overall_task], valid_data[diff_task])
        
        correlation_results.append({
            'Overall_Task': overall_task,
            'Difference_Task': diff_task,
            'Correlation': corr_coef,
            'P_Value': p_value,
            'N_Observations': len(valid_data),
            'Significant': 'Yes' if p_value < 0.05 else 'No'
        })
    else:
        correlation_results.append({
            'Overall_Task': overall_task,
            'Difference_Task': diff_task,
            'Correlation': np.nan,
            'P_Value': np.nan,
            'N_Observations': len(valid_data),
            'Significant': 'Insufficient data'
        })

# Convert to DataFrame
corr_df = pd.DataFrame(correlation_results)

# Display results
print("Correlations between Overall Task Measures and Difference Scores:")
print("=" * 70)
print(corr_df.round(4))

# Create a summary of significant correlations
significant_corrs = corr_df[corr_df['Significant'] == 'Yes']
if len(significant_corrs) > 0:
    print("\n\nSignificant Correlations (p < 0.05):")
    print("=" * 40)
    for idx, row in significant_corrs.iterrows():
        print(f"{row['Overall_Task']} vs {row['Difference_Task']}: "
              f"r = {row['Correlation']:.3f}, p = {row['P_Value']:.3f}")
else:
    print("\n\nNo significant correlations found (p < 0.05)")

# Alternative: Quick correlation calculation
print("\n\nQuick Correlation Summary:")
print("=" * 30)
for overall_task, diff_task in task_pairs:
    if overall_task in df_factor.columns and diff_task in df_factor.columns:
        corr = df_factor[overall_task].corr(df_factor[diff_task])
        print(f"{overall_task} vs {diff_task}: r = {corr:.3f}")
    else:
        print(f"{overall_task} vs {diff_task}: Missing column(s)")

# Optional: Create a heatmap visualization
try:
    import matplotlib.pyplot as plt
    import seaborn as sns
    
    # Create a subset DataFrame with only the task pairs
    task_columns = []
    for overall_task, diff_task in task_pairs:
        if overall_task in df_factor.columns and diff_task in df_factor.columns:
            task_columns.extend([overall_task, diff_task])
    
    if task_columns:
        task_subset = df_factor[task_columns]
        
        # Calculate correlation matrix
        corr_matrix = task_subset.corr()
        
        # Create figure
        plt.figure(figsize=(12, 10))
        
        # Create heatmap
        sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
                    square=True, fmt='.3f', cbar_kws={'label': 'Correlation'})
        
        plt.title('Correlation Matrix: Task Measures')
        plt.tight_layout()
        plt.show()
        
except ImportError:
    print("\nNote: Install matplotlib and seaborn for visualization")
    print("pip install matplotlib seaborn")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
from itertools import combinations

# Define the task pairs to correlate
task_pairs = [
    ('object_view_rt', 'object_view_rt_DD-DS'),
    ('object_view_acc', 'object_view_acc_DS-DD'),
    ('mapping_dist', 'mapping_dist_DD-DS'),
    ('dist_est_corr', 'dist_est_corr_DS-DD'),
    ('dist_est', 'dist_est_DD-DS'),
    ('dist_comp', 'dist_comp_DS-DD'),
    ('jrd_lr', 'jrd_lr_DS-DD'),
    ('jrd_diff', 'jrd_diff_DD-DS')
]

# Original targeted analysis
correlation_results = []
for overall_task, diff_task in task_pairs:
    valid_data = df_factor[[overall_task, diff_task]].dropna()
    
    if len(valid_data) > 2:
        corr_coef, p_value = pearsonr(valid_data[overall_task], valid_data[diff_task])
        
        correlation_results.append({
            'Overall_Task': overall_task,
            'Difference_Task': diff_task,
            'Correlation': corr_coef,
            'P_Value': p_value,
            'N_Observations': len(valid_data),
            'Significant': 'Yes' if p_value < 0.05 else 'No'
        })
    else:
        correlation_results.append({
            'Overall_Task': overall_task,
            'Difference_Task': diff_task,
            'Correlation': np.nan,
            'P_Value': np.nan,
            'N_Observations': len(valid_data),
            'Significant': 'Insufficient data'
        })

corr_df = pd.DataFrame(correlation_results)

print("Targeted Correlations between Overall Task Measures and Difference Scores:")
print("=" * 70)
print(corr_df.round(4))

# NEW: Comprehensive analysis of all variables in heatmap
# Create task_columns list (same as in original heatmap code)
task_columns = []
for overall_task, diff_task in task_pairs:
    if overall_task in df_factor.columns and diff_task in df_factor.columns:
        task_columns.extend([overall_task, diff_task])

# Remove duplicates while preserving order
task_columns = list(dict.fromkeys(task_columns))

print(f"\n\nAnalyzing all pairwise correlations for {len(task_columns)} variables:")
print("Variables included:", task_columns)
print("=" * 80)

# Calculate all pairwise correlations with significance testing
all_correlations = []

for var1, var2 in combinations(task_columns, 2):
    if var1 in df_factor.columns and var2 in df_factor.columns:
        # Get valid data for both variables
        valid_data = df_factor[[var1, var2]].dropna()
        
        if len(valid_data) > 2:
            corr_coef, p_value = pearsonr(valid_data[var1], valid_data[var2])
            
            all_correlations.append({
                'Variable_1': var1,
                'Variable_2': var2,
                'Correlation': corr_coef,
                'P_Value': p_value,
                'N_Observations': len(valid_data),
                'Significant': 'Yes' if p_value < 0.05 else 'No',
                'Effect_Size': 'Large' if abs(corr_coef) >= 0.5 else 
                              'Medium' if abs(corr_coef) >= 0.3 else 'Small'
            })

# Convert to DataFrame
all_corr_df = pd.DataFrame(all_correlations)

# Display all significant correlations
significant_all = all_corr_df[all_corr_df['Significant'] == 'Yes'].copy()

if len(significant_all) > 0:
    # Sort by absolute correlation strength
    significant_all['Abs_Correlation'] = abs(significant_all['Correlation'])
    significant_all = significant_all.sort_values('Abs_Correlation', ascending=False)
    
    print(f"\nALL SIGNIFICANT CORRELATIONS (p < 0.05):")
    print("=" * 50)
    print(f"Found {len(significant_all)} significant correlations out of {len(all_corr_df)} total pairs")
    print()
    
    for idx, row in significant_all.iterrows():
        direction = "positive" if row['Correlation'] > 0 else "negative"
        print(f"{row['Variable_1']} ↔ {row['Variable_2']}")
        print(f"  r = {row['Correlation']:.3f} ({direction}), p = {row['P_Value']:.4f}")
        print(f"  Effect size: {row['Effect_Size']}, N = {row['N_Observations']}")
        print()
    
    # Summary by effect size
    print("\nSUMMARY BY EFFECT SIZE:")
    print("=" * 25)
    effect_summary = significant_all['Effect_Size'].value_counts()
    for effect_size, count in effect_summary.items():
        print(f"{effect_size} effects (|r| ≥ {0.5 if effect_size=='Large' else 0.3 if effect_size=='Medium' else 0.0}): {count}")
    
    # Display strongest correlations
    print(f"\nSTRONGEST CORRELATIONS (Top 5):")
    print("=" * 30)
    top_corrs = significant_all.head(5)
    for idx, row in top_corrs.iterrows():
        print(f"r = {row['Correlation']:.3f}: {row['Variable_1']} ↔ {row['Variable_2']}")

else:
    print("\nNo significant correlations found among all variable pairs (p < 0.05)")

# Create enhanced correlation matrix with significance indicators
print(f"\n\nCORRELATION MATRIX WITH SIGNIFICANCE:")
print("=" * 40)

if task_columns:
    task_subset = df_factor[task_columns]
    corr_matrix = task_subset.corr()
    
    # Create significance matrix
    n_vars = len(task_columns)
    sig_matrix = np.full((n_vars, n_vars), '', dtype=object)
    
    for i, var1 in enumerate(task_columns):
        for j, var2 in enumerate(task_columns):
            if i != j:  # Skip diagonal
                valid_data = df_factor[[var1, var2]].dropna()
                if len(valid_data) > 2:
                    _, p_val = pearsonr(valid_data[var1], valid_data[var2])
                    if p_val < 0.001:
                        sig_matrix[i, j] = '***'
                    elif p_val < 0.01:
                        sig_matrix[i, j] = '**'
                    elif p_val < 0.05:
                        sig_matrix[i, j] = '*'
    
    # Display matrix with significance
    print("Correlation coefficients with significance markers:")
    print("* p < 0.05, ** p < 0.01, *** p < 0.001")
    print()
    
    for i, var1 in enumerate(task_columns):
        print(f"{var1}:")
        for j, var2 in enumerate(task_columns):
            if i != j:
                corr_val = corr_matrix.iloc[i, j]
                sig_marker = sig_matrix[i, j]
                print(f"  vs {var2}: {corr_val:.3f}{sig_marker}")
        print()

# Optional: Create enhanced heatmap visualization
try:
    import matplotlib.pyplot as plt
    import seaborn as sns
    
    if task_columns:
        task_subset = df_factor[task_columns]
        corr_matrix = task_subset.corr()
        
        # Create significance mask
        mask = np.zeros_like(corr_matrix, dtype=bool)
        for i, var1 in enumerate(task_columns):
            for j, var2 in enumerate(task_columns):
                if i != j:
                    valid_data = df_factor[[var1, var2]].dropna()
                    if len(valid_data) > 2:
                        _, p_val = pearsonr(valid_data[var1], valid_data[var2])
                        if p_val >= 0.05:
                            mask[i, j] = True
        
        plt.figure(figsize=(14, 12))
        
        # Create heatmap with significance masking
        sns.heatmap(corr_matrix, 
                    annot=True, 
                    cmap='coolwarm', 
                    center=0,
                    square=True, 
                    fmt='.3f', 
                    cbar_kws={'label': 'Correlation'},
                    mask=mask,  # Mask non-significant correlations
                    linewidths=0.5)
        
        plt.title('Correlation Matrix: Task Measures\n(Only significant correlations shown, p < 0.05)')
        plt.xticks(rotation=45, ha='right')
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()
        
except ImportError:
    print("\nNote: Install matplotlib and seaborn for enhanced visualization")
    print("pip install matplotlib seaborn")
```
